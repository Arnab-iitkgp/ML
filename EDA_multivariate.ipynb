Bivariate and multivariate

```python
import pandas as pd
import seaborn as sns
```

```python
tips = sns.load_dataset("tips") # inbuilt in the sns lib
```

```python
flights = sns.load_dataset("flights")
iris  = sns.load_dataset("iris")
```

```python
titanic = pd.read_csv('/content/Titanic-Dataset.csv')
```

```python
tips.head()
```

```python
#scatterplot (numerical - numercal) bivariate analysis
sns.scatterplot(x='total_bill', y='tip', data=tips)
```

```python
#multivariate ,   IMPORTANT
sns.scatterplot(x='total_bill', y='tip', hue ='sex', style = 'smoker', size = 'size', data=tips)
```

**BAR PLOT**

```python
titanic.head()
```

```python
mean_fare_pclass1 = titanic[titanic['Pclass'] == 1]['Fare'].mean()
print(f"Mean Fare for Pclass 1: {mean_fare_pclass1}")
```

```python
# sns.barplot(x='Pclass', y='Fare',data=titanic) # getting average at top point or edge
sns.barplot(x='Pclass', y='Fare',hue ='Sex',data=titanic)
```

```python
sns.boxplot(x='Sex', y='Age',hue='Survived',data = titanic)
```

```python
#probability distribution
sns.distplot(titanic[titanic['Survived']==0]['Age'],hist=False) #died
sns.distplot(titanic[titanic['Survived']==1]['Age'],hist=False) #survived

#one obs is that for younger ages probab of dying is less , which is infactin real true.
# so the beauty is that we are getting quite knowledge from the data.
```

```python
# Heatmap
titanic.head(3)
```

```python
pd.crosstab(titanic['Pclass'],titanic['Survived'])
```

```python
#categorical with categorical
sns.heatmap((pd.crosstab(titanic['Pclass'],titanic['Survived'])))
```

```python
titanic.groupby('Pclass')['Survived'].mean()*100
```
